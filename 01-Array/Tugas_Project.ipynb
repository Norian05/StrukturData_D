```python
import matplotlib.pyplot as plt
import numpy as np

# Defenisikan array kosong
array = []

# Looping untuk input data
for i in range(10):
  nilai = int(input("Input nilai Mahasiswa: "))
  array.append(nilai)
```

```python
# Menampilkan nilai tertinggi, terendah, dan rata-rata
print("Nilai Tertinggi Mahasiswa:", max(array))
print("Nilai Terendah Mahasiswa :", min(array))
print("Nilai Rata-Rata Mahasiswa:", sum(array)/len(array))
```

```python
# Perhitungan hasil input
lulus = 0
tidak_lulus = 0
for nilai in array:
  if nilai >=60:
    lulus += 1
  else:
    tidak_lulus += 1

# Menampilkan hasil dari perhitungan
print("Mahasiswa Lulus      :", lulus)
print("Mahasiswa Tidak Lulus:", tidak_lulus)
```

```python
plt.figure(figsize=(8, 5))
plt.bar(range(len(array)), array)

plt.title("Nilai Seluruh Mahasiswa", fontsize=14)
plt.xlabel("Mahasiswa", fontsize=12)
plt.ylabel("Nilai", fontsize=12)
plt.ylim(0, 100)
plt.grid(axis="y", linestyle="--", alpha=0.7)
plt.show()
```

```python
kategori =["Lulus", "Tidak Lulus"]
jumlah = [lulus, tidak_lulus]
plt.figure(figsize=(8, 5))
plt.bar(kategori, jumlah, color=["blue", "red"])

plt.title("Data Kelulusan", fontsize=15)
plt.ylabel("Jumlah Mahasiswa", fontsize=12)
plt.grid(axis="y", linestyle="--", alpha=1)
plt.ylim(0, 10)
plt.show()
```
